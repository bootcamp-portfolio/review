#### map함수(list&함수)<<<<>>>>
:여러개의 데이터(이터러블 리스트등)에 같은 함수를 한번에 적용해주는 함수
-반복문으로 하나씩 처리하던 작업을 한줄로 간단하게 표현
-format : map(함수,반복가능한 자료형)
-list(map(함수,list))로 많이 사용(for x)

```python
###for와 map의 비교
#for: 제공된 데이터에 2를 곱하여 새로운 List만들기

numbers=[1,2,3,4]
result=[]

result=[i*2 for i in numbers]

print(result)
```

```python
result=[]

for n in numbers:
    result.append(n*2)
print(result)
```

```python
#map: lambda함수로 제공된 데이터에 2를 곱하여 새로운 List만들기
#map 병렬처리 한번에 빠름
numbers=[1,2,3,4]
result=list(map(lambda x:x*2,numbers))
result
```

```python
#map: 일반함수로 제공된 데이터에 2를 곱하여 새로운 List만들기(3.10부터)
numbers=[1,2,3,4]
def numSquare(n):
    return n*2

result=list(map(numSquare,numbers))
#함수 괄호 없음,x도 없음
#병렬처리->멀티프로세스,멀티 쓰레드
result
```

```python
#for문:문자열을 대문자로 변환
words=['apple','banana','cherry']
result=[]

for w in words:
    result.append(w.upper())

print(result)
```

```python
#map: 함수를 이용하여 문자열을 대문자로 변환
words=['apple','banana','cherry']

def convertUpper(n):
    return n.upper()

result=list(map(str.upper,words))#글자 함수 str
result2=list(map(lambda x: x.upper(),words))#글자 함수 str
result3=list(map(convertUpper,words))
print(result)
print(result2)
print(result3)
```

```python
#숫자열을 문자열로 변환
#for문 
nums=[1,2,3]

def convertString(n):
    return str(n)

#map내장함수
result=list(map(str,nums))##
#map람다함수
result2=list(map(lambda x: str(x),nums))
#map제작함수
result3=list(map(convertString,nums))

print(result)
print(result2)
print(result3)
```

```python
#class (분 머신 딥-케라스-구글?)
# 파이토치:페이스북-모듈(자연어처리):변화가 쉬워 많이 사용

class Add():
    #property
    num1=0
    num2=0

    #constructor
    def __init__(self, num1, num2): #본인 자신?
        self.num1= num1
        self.num2=num2

    #method
    def add(self): #this->self
        result=self.num1+self.num2
        return result #class
```

```python
addition= Add(1,2)#생성자 인스턴스?
print(addition.add())
```

```python
###########################
#addition1=Add()
#addition.num1=10
#addition.num2=20
#addition1.add()
#############################
```

```python
class Sub(Add):
    num1=0
    num2=0

    def __init__(self,num1,num2):
        self.num1=num1
        self.num2=num2
    def sub(self):
        result=self.num1-self.num2
        return result
```

```python
sub= Sub(2,3)#상속
print(sub.add())
print(sub.sub())
```
